## Training model

```python
from ultralytics import YOLO
import torch
from pathlib import Path
import os
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Checking for GPU with CUDA
device = 0 if torch.cuda.is_available() else 'cpu'

model = YOLO('yolo26n.pt') 
```

```python
project_path = Path('~/Desktop/uav-edge-intelligence').expanduser()
```

```python
yaml_path = Path('~/.cache/kagglehub/datasets/uav-edge-intelligence/data.yaml').expanduser()

results = model.train(
    data=yaml_path,
    epochs=1,
    imgsz=640,
    batch=16,
    device=device,
    project=project_path,
    name='uavEdgeIntelligence'
)
```
