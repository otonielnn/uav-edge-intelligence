## Training model

```python
from ultralytics import YOLO
import torch
from pathlib import Path
```

```python
# Checking for GPU with CUDA
device = 0 if torch.cuda.is_available() else 'cpu'

model = YOLO("yolo11n.pt")
```

```python
yaml_path = Path('~/.cache/kagglehub/datasets/uav-edge-intelligence/data.yaml').expanduser()

train = model.train(
    data=yaml_path,
    epochs=50,
    imgsz=640,
    batch=4,
    patience=20,
)
```
